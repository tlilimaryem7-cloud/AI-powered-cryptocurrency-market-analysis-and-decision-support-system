### Example to interpret the predection :

Example:
  P(DOWN) = 0.2946  →  29.46%

  Direction  = UP

  | Confidence | Meaning |
|------------|----------|
| 50% | Model has no idea  equivalent to a coin flip |
| 55–60% | Weak signal  low conviction |
| 60–70% | Moderate signal  some predictive edge |
| 70–80% | Strong signal act with reasonable conviction |
| 80%+ | Very strong signal rare, high conviction prediction |
